```python
import ultralytics
ultralytics.checks()
```

```python
from ultralytics import YOLO

model = YOLO("yolov8s.pt")
```

```python
results = model.train(
    data="/content/dataset/data.yaml",
    epochs=50,
    patience=10,
    imgsz=640,
    batch=16,
    optimizer="AdamW",
    lr0=0.001,
    name='rs_ddd_2classes_yolov8s',
    project='/content/drive/MyDrive/graduation_project/dataset',
    save_period=10,
    device=0
)
```
