```python
import os
import cv2
import matplotlib.pyplot as plt
```

```python
folder = "/content/eyes_dataset/Dataset/Eyes_dataset"
classes = ['Open_Eyes', 'Closed_Eyes']
splits = ['train', 'val', 'test']
```

```python
counts = {s: [len(os.listdir(os.path.join(folder, s, c))) for c in classes]
          for s in splits}

plt.figure(figsize=(8,5))
x = range(len(splits))

for i, cls in enumerate(classes):
    plt.bar([p + i*0.25 for p in x],
            [counts[s][i] for s in splits],
            width=0.23, label=cls)

    for j, split in enumerate(splits):
        plt.text(j + i*0.25, counts[split][i] + 50,
                 str(counts[split][i]), ha='center', fontsize=11)

plt.xticks([p + 0.125 for p in x], splits)
plt.ylabel("Số lượng ảnh")
plt.title("Phân bố nhãn – Tập dữ liệu mắt")
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
from ultralytics import YOLO

# model = YOLO("yolov8s-cls.pt")
model = YOLO("yolov8n-cls.pt")
```

```python
model.train(
    data="/content/eyes_dataset/Dataset/Eyes_dataset",
    epochs=50,
    imgsz=96,
    batch=64,

    fliplr=0.5,
    degrees=3,
    translate=0.08,
    scale=0.15,
    erasing=0.2,

    optimizer="AdamW",
    lr0=0.001,
    patience=10,
    cos_lr=True
)
```

```python
from ultralytics import YOLO

model = YOLO("/content/runs/classify/train/weights/best.pt")

print(model.names)
```

```python
result = model.predict("/content/demoImg.webp")[0]

img_plot = result.plot()

plt.figure(figsize=(6,6))
plt.imshow(cv2.cvtColor(img_plot, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```
